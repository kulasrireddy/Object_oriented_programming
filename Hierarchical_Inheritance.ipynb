### Hierarchical Inheritance

```python
#one base class and more than one derived classes are acquiring properties from base class.
```

```python
class Parent:
    def pclass(self):
        print("PARENT CLASS")
class Son(Parent):
    def sclass(self):
        print("SON CLASS")
class Daughter(Parent):
    def dclass(self):
        print("DAUGHTER CLASS")
p = Parent()
```

```python
p.pclass()
```

```python
p.sclass()
```

```python
s = Son()
```

```python
s.sclass()
```

```python
s.pclass()
```

```python
s.dclass()
```

```python
d = Daughter()
```

```python
d.dclass()
```

```python
d.pclass()
```

```python
d.sclass()
```

```python
class Student:
    def __init__(self, name, age, college, degree,cgpa):
        self.name = name
        self.age = age
        self.college = college
        self.degree = degree
        self.cgpa = cgpa
    def display_student(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("CGPA:",self.cgpa)
    def is_eligible(self):
        if self.cgpa >= 7.0:
            return True
        else:
            return False
class PlacementStudent(Student):
    def __init__(self,name, age, college, degree, cgpa, skills= None, dsa_score = 0):
        super().__init__(name, age, college,degree, cgpa)
        self.skills = skills
        self.dsa_score = dsa_score
    def add_skill(self, skill):
        self.skills = self.skills if self.skills is not None else []
        self.skills.append(skill)
        print("Added Successfully")
    def check_placement(self):
        if self.cgpa >= 7.0 and self.dsa_score >= 60:
            return True
        else:
            return False
    def display_placement_profile(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("CGPA:",self.cgpa)
        print("Skills:",self.skills)
        print("DSA_Score:",self.dsa_score)
class HigherStudiesStudent(Student):
    def __init__(self,name, age, college, degree, cgpa, exam, exam_score = 0):
        super().__init__(name, age, college, degree, cgpa)
        self.exam = exam
        self.exam_score = exam_score
    def check_higher_studies_student(self):
        if self.cgpa >= 7.0 and self.exam_score >= 50:
            return True
        else:
            return False
    def display_higher_student_profile(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("CGPA:",self.cgpa)
        print("Exam:",self.exam)
        print("Exam Score:",self.exam_score)
p1 = PlacementStudent(
    "Deepika",
    23,
    "Aditya University",
    "MCA",
    9.08,
    ["Python", "SQL"],
    75
)
    
```

```python
p1.display_student()

print(p1.is_eligible())

p1.add_skill("JavaScript")

p1.display_placement_profile()

print(p1.check_placement())
```
