## Polymorphism

```python
#Polymorphism means the implementing the same thing in diff ways
#Two types of polymorphism
#1.Complie time (Achieve by using methodoverloading)
#2.Run time  (Achieve by using methodoverriding)
```

```python
#Complie time (take recent method only)
class Solution:
    def add(self,a,b):
        print(a+b)
    def add(self,a, b, c):
        print(a+b+c)
    def add(self,a, b, c=10):
        print(a+b+c)
a = Solution()
a.add(100,100)
```

```python
class Demo: #methodoverloading by using default arguemets
    def sub(self,a,b):
        print(a-b)
d = Demo()
d.sub(10,20)
```

```python
class Operations:  #methodoverloading by using positional arguments
    def div(self,*args):   #methodoverloading can solve also with **kwargs
        print(args[0]/args[1])
o = Operations()
o.div(10,5)
```

```python
#Run-time( we achieve this by using overridiing concept 
class Base:    #methodoverriding means same method names, and same parameters in both parent class and dervied class
    def transport(self):
        print("Rikshaw")
class Derived:
    def transport(self):
        print("Bus")
v = Base()
d = Derived()
v.transport()
d.transport()
```

```python
#Run-time
class Payment:
    def pay(self):
        print("Payment successfully Completed")
class UPIPayment(Payment):
    def pay(self):
        print("UPI Payment Completed")
class CardPayment(Payment):
    def pay(self):
        print("Card Payment Completed")
class CashPayment(Payment):
    def pay(self):
        print("Cash Payment Completed")
u = UPIPayment()
c = CardPayment()
p = CashPayment()
u.pay()
c.pay()
p.pay()
```

```python
payments = [UPIPayment(),CardPayment(),CashPayment()]
for p in payments:
    p.pay()
```
