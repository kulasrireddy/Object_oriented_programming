### Multilevel Inheritance

```python
# one baseclass and one derived class, and that dervied class acts as baseclass for other dervied class
```

```python
class GrandParent:
    def gpclass(self):
        print("GRANDPARENT CLASS")
class Parent(GrandParent):
    def pclass(self):
        print("PARENT CLASS")
class Child(Parent):
    def cclass(self):
        print("CHILD CLASS")

#OBJ FOR CHILD
c = Child()
c.cclass()
c.pclass()
c.gpclass()
```

```python
class Person:
    def __init__(self,name, age):
        self.name = name
        self.age = age
    def display_person(self):
        print("Name:",self.name)
        print("Age:",self.age)
class Student(Person):
    def __init__(self, name, age, college, degree):
        super().__init__(name, age)
        self.college = college
        self.degree = degree
    def display_student(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("College:",self.college)
        print("Degree:",self.degree)
    def is_adult(self):
        if self.age >= 18:
            return True
        else:
            return False
class PlacementStudent(Student):
    def __init__(self,name, age, college, degree, skills = None, dsa_score= 0):
        super().__init__(name, age, college, degree)
        self.skills = skills
        self.dsa_score = dsa_score
    def add_skill(self, skill):
        self.skills = self.skills if self.skills is not None else []
        self.skills.append(skill)
        print("Skill added Successfully")
    def check_placement(self):
        if self.dsa_score > 60 and self.is_adult() == True:
            return "Eligible for Placement"
        else:
            return "No
    def display_profile(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("Skills:",self.skills)
        print("DSA_Score:",self.dsa_score)

#Child class
p1 = PlacementStudent(
    "Deepika",
    23,
    "Aditya University",
    "MCA",
    ["Python", "SQL"],
    75
)
```
