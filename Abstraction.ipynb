## Abstraction

```python
#Abstraction means hiding the implementation, only display the essentional part
#Abstraction Class means a class haivng atleast one abstract method
#Abstract method means function calling but empty definition, no logic only fun call
#decorator (@) helps to implement the abstract method in abstract class
#for Python "abc" is module for abstraction and "ABC" is abstract class to class to behave like a abstract class
#object creation is not possible for abstract class.
#To create object we need to convert the abstract class into concrete class, concrete class means should have all methods what abstract class have but if it missied one method then it is also abstract class
#Concrete class means the derived class who overrides the methods in abstract class.it can create obj to show the methods in abstarct class, because in abstract class creating obj is not possible.
```

```python
from abc import ABC, abstractmethod
class AbstractClass(ABC):   #ABSTRACT CLASS
    @abstractmethod
    def UPIPayment(self):   #ABSTRACT METHOD1
        None
    @abstractmethod
    def CardPayment(self):  #ABSTRACT METHOD2
        None
    @abstractmethod
    def CashPayment(self):   #ABSTRATC METHOD3
        None
class Loan(AbstractClass):   #CONCRETE CLASS
    def UPIPayment(self):    #CONCRETE METHOD1
        print("UPI Payment Done!")
    def CardPayment(self):   #CONCRETE METHOD2
        print("Card Payment Done!")
    def CashPayment(self):    #CONCRETE METHOD3
        print("Cash Payment Done!")
l = Loan()
l.UPIPayment()
l.CardPayment()
l.CashPayment()
```

```python
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
    def payment_success(self):
        print("Payment Successfully")
class UPIPayment(Payment):
    def pay(self, amount):
        self.amount = amount
    def payment_success(self):
        print(f"Payment {self.amount} UPI Successfully")
class CardPayment(Payment):
    def pay(self,amount):
        self.amount= amount
    def payment_success(self):
        print(f"Payment {self.amount} Card Successfully")
c = CardPayment()
c.pay(500)
c.payment_success()
u = UPIPayment()
u.pay(900)
u.payment_success()
```
