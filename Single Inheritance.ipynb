## Inheritance

```python
class Animal:
    def Dog(self):
        print("Barking")
    def Cat(self):
        print("Mewom")
class Sound(Animal):
    def Bride(self):
        print("Slow sound")
    def Female_cat(self):
        print("High sound")
d = Sound()
d.Bride()
d.Dog()
d.Cat()
```

### Single Inheritance

```python
class Student:
    def __init__(self,name,college, degree, cgpa):
        self.name = name
        self.college = college
        self.degree = degree
        self.cgpa = cgpa
    def display_std(self):
        print("Name:",self.name)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("CGPA:",self.cgpa)
    def is_eligible(self):
        if self.cgpa >= 7.0:
            print("Eligible")
class PlacementStudent(Student):
    def __init__(self, name, college, degree, cgpa,skills= None, dsa_score=0):  # Child constructor
        super().__init__(name, college, degree, cgpa)                           #accessing the parent class
        self.skills = skills
        self.dsa_score = dsa_score
    def add_skill(self,skill):
        self.skills = self.skills if self.skills is not None else []
        self.skills.append(skill)
        print("Skill added Successfully")
    def check_placment_eligibility(self):
        if self.cgpa >= 7.0 and self.dsa_score >= 60:
            print("You are eligible for Placement")
    def display_placement_profile(self):
        print("Name:",self.name)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("CGPA:",self.cgpa)
        print("Skills:",self.skills)
        print("DSA_Score:",self.dsa_score)
p1 = PlacementStudent(
    "Deepika",
    "Aditya University",
    "MCA",
    9.08,
    ["Python", "SQL"],
    75
)
```

```python
p1.check_placment_eligibility()
```

```python
p1.display_std()
```

```python
p1.display_placement_profile()
```
