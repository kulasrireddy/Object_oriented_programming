## Encapsulation

```python
#Encapsulation means wrapping data and methods together inside a class + restricting direct access to data
```

```python
#To hide the methods and varibles we need to access specifier to restrict the data within class
#Access Specifiers are 
#1. Public
#2. Private(use "__"(two underscores) before method or varibles then it is treated as private data)
#Private is only access within class only, Public can access both inside or outside of class
```

```python
class Demo:
    a = 100
    b = 200
    def display(self):
        print("The data is displayed")
d = Demo()
print(d.a,d.b)
d.display()
```

```python
class Demo:
    __a = 100   #priavte varible
    b = 200
    def __display(self):  #priavte method
        print("The data is displayed")
d = Demo()
print(d.a,d.b)
d.display()
```

```python
class Demo:
    __a = 100   #priavte varible
    b = 200
    def display(self):
        print(self.__a)  #accessing inside the class of priavte varible
        print("The data is displayed")
d = Demo()
print(d.b)
d.display()
```

```python
class Demo:
    __a = 100   #priavte varible
    b = 200
    def __display(self): #priavte method   
        print(self.__a)
        print("The data is displayed")
    def show(self):  #accessing within class of the priavte method
        self.__display()
d = Demo()
print(d.b)
d.show()
```

```python
class BankAccount:
    def __init__(self, account_holder, balance=10000):
        self.account_holder = account_holder
        self.__balance = balance
    def deposit(self, amount):
        if amount <= 0:
            print("Invalid Amount")
        self.__balance += amount
        print("Amount Deposited Successfully")
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid Amount")
        if amount > self.__balance:
            print("Insufficient Balance")
        else:
            print("You can Withdraw it")
            self.__balance -= amount
    def get_balance(self):
        print("Balance:",self.__balance)
b = BankAccount(123)
b.deposit(500)
b.get_balance()
b.withdraw(5000)
b.get_balance()
# d.deposit(-500)
```
