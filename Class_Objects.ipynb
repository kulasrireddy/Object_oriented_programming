### Task1

```python
class Student:
    def __init__(self,name,age,course,cgpa):
        self.name = name
        self.age = age
        self.course = course
        self.cgpa = cgpa
    def display_details(self):
        print("Name:",self.name)
        print("Age:",self.age)
        print("Course:",self.course)
        print("CGPA:",self.cgpa)
    def is_eligible(self):
        if self.cgpa >= 7.0:
            print("Eligible")
        else:
            print("Not Eligible")
s1 = Student("Deepika",23,"MCA",9.08)
s2 = Student("Ganesha",26,"PHD",9.99)
            
            
```

```python
s1.display_details()
```

```python
s1.is_eligible()
```

```python
s2.display_details()
```

```python
s2.is_eligible()
```

### Task2

```python
class BankAccount:
    def __init__(self, acc_holder,acc_number,balance=0):
        self.acc_holder = acc_holder
        self.acc_number = acc_number
        self.balance = balance
    def display_balance(self):
        print("Balance:",self.balance)
    def deposit(self,amount):
        self.balance += amount
        print("Amount deposited Successfully")
    def withdraw(self,amount):
        if self.balance < amount:
            print("Insufficient Balance")
        else:
            self.balance -= amount
            print("Amount Withdraw Successfully")
b1 = BankAccount("Deepika",1234)            
            
```

```python
b1.deposit(60000)
```

```python
b1.display_balance()
```

```python
b1.withdraw(20000)
```

```python
b1.display_balance()
```

```python
b1.withdraw(50000)
```

### Task3

```python
class Employee:
    def __init__(self,name,emp_id,department,salary):
        self.name = name
        self.emp_id = emp_id
        self.department = department
        self.salary = salary
    def display_details(self):
        print("Name:",self.name)
        print("Employee ID:",self.emp_id)
        print("Department:",self.department)
        print("Salary:",self.salary)
    def annual_salary(self):
        annual_salary = self.salary * 12
        return annual_salary
    def hike(self,amount):
        self.salary += amount
        return self.salary
e1 = Employee("Deepika",5,"IT",75000)  
```

```python
e1.display_details()
```

```python
e1.annual_salary()
```

```python
e1.hike(15000)
```

```python
e1.annual_salary()
```

```python
print(e1.annual_salary())
```

### Task4

```python
class Rectangle:
    def __init__(self,l,w):
        self.l = l
        self.w = w
    def area(self):
        a = self.l * self.w
        return a
    def perimeter(self):
        p = 2 * (self.l + self.w)
        return p
    def display_details(self):
        print("Area:",self.area())
        print("Perimeter:",self.perimeter())
    def is_square(self):
        return self.l == self.w
r1 = Rectangle(5,2)
r2 = Rectangle(2,2)

# self.area = method itself
# self.area() = execute method and get its returned value
```

```python
r1.area()
```

```python
r1.perimeter()
```

```python
r1.display_details()
```

```python
r2.is_square()
```

```python
r1.is_square()
```

```python
class Placement_List:
    def __init__(self,name,college,degree,cgpa,skills=None,dsa_score=0):
        self.name = name
        self.college = college
        self.degree = degree
        self.cgpa  = cgpa
        self.skills = skills
        self.dsa_score = dsa_score
    def display_profile(self):
        print("Name:",self.name)
        print("College:",self.college)
        print("Degree:",self.degree)
        print("CGPA:",self.cgpa)
        print("Skills:",self.skills)
        print("DSA Score:",self.dsa_score)
    def add_skill(self,skill):
        self.skills = self.skills if self.skills is not None else []
        self.skills.append(skill)
        print("Added Successfully")
    def show_skills(self):
        print("Skills:",self.skills)
    def check_eligibility(self):
        return self.cgpa >= 7.0 and self.dsa_score >= 60
p1 = Placement_List(
    "Deepika",
    "Aditya University",
    "MCA",
    9.08,
    ["Python", "SQL"],
    75
)        
p2 = Placement_List(
    "Deepu",
    "Aditya University",
    "MCA",
    9.08,
    ["SQL"],
    85
)        
```

```python
p1.check_eligibility()
```

```python
p1.add_skill("HTML")
```

```python
p1.show_skills()
```

```python
p1.display_profile()
```

```python
p2.display_profile()
```

```python
p2.check_eligibility()
```

```python
p1.show_skills()
```

```python
p2.show_skills()
```
