### Multiple Inheritance

```python
#one derived class is acquiring properties from multiple base classes
```

```python
class Father:
    def fclass(self):
        print("Father Class")
class Mother:
    def mclass(self):
        print("Mother Class")
class Child(Father, Mother):
    def cclass(self):
        print("Child Class")
```

```python
c = Child()
```

```python
c.cclass()
```

```python
c.fclass()
```

```python
c.mclass()
```

```python
class Student:
    def __init__(self,name, rollno, age, college):
        self.name = name
        self.rollno = rollno
        self.age = age
        self.college = college
    def display_Profile(self):
        print("Name:",self.name)
        print("Rollno:",self.rollno)
        print("Age:",self.age)
        print("College:",self.college)
class Skills:
    def __init__(self, skills= None):
        self.skills = skills
    def display_skills(self):
        print("Skills:",self.skills)
    def add_skills(self,skill):
        self.skills = self.skills if self.skills is not None else []
        self.skills.append(skill)
        print("Added Successfully")
class Placement:
    def __init__(self,company, package):
        self.company = company
        self.package = package
    def display_placement(self):
        print("Company:",self.company)
        print("Package:",self.package)
class PlacementStudent(Student, Skills, Placement):
    def __init__(self,name, rollno, age, college,skills,company, package,status):
        Student.__init__(self,name, rollno, age, college)      #ClassName.__init__() = specific parent ni direct ga call cheyyadam.
        Skills.__init__(self,skills)                           #super().__init__() = MRO prakaram next class ni call cheyyadam(Method Resolution Order.) 
        Placement.__init__(self,company, package)              #Python oka method/attribute kosam classes ni ye order lo search cheyyalo decide chese order = MRO.
        self.status = status
    def display_profile(self):
        print("Name:",self.name)
        print("Rollno:",self.rollno)
        print("Age:",self.age)
        print("College:",self.college)
        print("Skills:",self.skills)
        print("Company",self.company)
        print("Package:",self.package)
        print("Status:",self.status)
p1 = PlacementStudent(
    "Deepika",
    16,
    23,
    "Aditya University",
    ["Python", "SQL"],
    "Accenture",
    700000,
    "Selected"
)

p1.display_profile()
        
        
        
```
